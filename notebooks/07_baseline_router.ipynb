````python
import ast

source = V71_FILE.read_text(
    encoding="utf-8"
)

tree = ast.parse(
    source,
    filename=str(V71_FILE)
)

top_level = []

for node in tree.body:

    if isinstance(node, ast.ClassDef):
        top_level.append(
            f"class {node.name}"
        )

    elif isinstance(node, ast.FunctionDef):
        top_level.append(
            f"def {node.name}"
        )

    elif isinstance(
        node,
        ast.AsyncFunctionDef
    ):
        top_level.append(
            f"async def {node.name}"
        )

    elif isinstance(
        node,
        ast.Import
    ):
        top_level.append(
            "import"
        )

    elif isinstance(
        node,
        ast.ImportFrom
    ):
        top_level.append(
            f"from {node.module or ''}"
        )


print("=" * 100)
print("V7.1 SOURCE ANALYSIS")
print("=" * 100)

for item in top_level:
    print(item)

print(
    "\nV71Router found in source:",
    "class V71Router" in source
)
````
